```python

import pandas as pd
import matplotlib.pyplot as plt

# Illustrative daily strategy returns (₹), not actual trading results
daily_pnl = [500, -200, 350, 150, -100, 600, -250, 400, 200, -150]

df = pd.DataFrame({
    "Day": range(1, len(daily_pnl) + 1),
    "Daily_PnL": daily_pnl
})

df["Cumulative_PnL"] = df["Daily_PnL"].cumsum()
df["Equity"] = 10000 + df["Cumulative_PnL"]

print("Illustrative total P&L:", df["Daily_PnL"].sum())
print("Illustrative number of days:", len(df))

plt.figure(figsize=(10, 5))
plt.plot(df["Day"], df["Equity"], marker="o")
plt.title("Illustrative Equity Curve — Sample Data")
plt.xlabel("Day")
plt.ylabel("Equity (₹)")
plt.grid(True)
plt.show()

df
```
