```python

import pandas as pd
import matplotlib.pyplot as plt

# Angel One monthly Net Realised P&L (₹)
data = {
    "Month": [
        "Oct 2025",
        "Nov 2025",
        "Dec 2025",
        "Feb 2026",
        "May 2026",
        "Jul 2026"
    ],
    "Monthly_PnL": [
        1421.47,
        375.73,
        10788.94,
        -217.04,
        -197.02,
        13251.29
    ]
}

df = pd.DataFrame(data)
df["Cumulative_PnL"] = df["Monthly_PnL"].cumsum()

print("MONTHLY ANGEL ONE P&L")
display(df)
print(f"Total of displayed monthly P&L: ₹{df['Monthly_PnL'].sum():.2f}")

# Monthly P&L chart
plt.figure(figsize=(11, 5))
plt.bar(df["Month"], df["Monthly_PnL"])
plt.axhline(0, linewidth=0.8)
plt.title("Angel One Monthly Net Realised P&L")
plt.xlabel("Month")
plt.ylabel("Net P&L (₹)")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("monthly_pnl.png", dpi=150)
plt.show()

# Cumulative reported P&L chart
plt.figure(figsize=(11, 5))
plt.plot(
    df["Month"],
    df["Cumulative_PnL"],
    marker="o"
)
plt.title("Cumulative Monthly Reported P&L")
plt.xlabel("Month")
plt.ylabel("Cumulative P&L (₹)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("monthly_cumulative_pnl.png", dpi=150)
plt.show()

df.to_csv("monthly_pnl.csv", index=False)
```
