```python

import pandas as pd
import matplotlib.pyplot as plt

# Actual values from Angel One Tax P&L reports
# FY 2025-26: Equity net P&L + F&O net P&L
# FY 2026-27: Equity net P&L + F&O net P&L

results = pd.DataFrame({
    "Period": ["FY 2025-26", "FY 2026-27 (to date)"],
    "Equity_Net_PnL": [12523.76, 13054.25],
    "F&O_Net_PnL": [-154.65, 0.00]
})

results["Combined_Net_PnL"] = (
    results["Equity_Net_PnL"] + results["F&O_Net_PnL"]
)
results["Cumulative_Reported_PnL"] = results["Combined_Net_PnL"].cumsum()

print("Angel One reported P&L summary")
display(results)

print(f"Combined reported P&L: ₹{results['Combined_Net_PnL'].sum():.2f}")

plt.figure(figsize=(9, 5))
plt.plot(
    results["Period"],
    results["Cumulative_Reported_PnL"],
    marker="o"
)
plt.title("Cumulative Reported P&L by Financial Year")
plt.xlabel("Reporting period")
plt.ylabel("Cumulative P&L (₹)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
